```python
import pandas as pd

# 读取影评文件
with open(r"C:\Users\45713\Desktop\pythonneed\movie_review.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
# 按 `=====` 分割影评
reviews = raw_text.split("=====")

# 清理数据并转换为 DataFrame
reviews = [review.strip() for review in reviews if review.strip()]
df = pd.DataFrame({"text": reviews})

# 确保数据格式正确
print(df.head())
```

```python
from datasets import Dataset

# 转换为 Dataset 格式
dataset = Dataset.from_pandas(df)

# 确保数据加载正确
print(dataset[0])  # 应该输出 {"text": "Movie Review: Dune ..."}
```

```python
from transformers import AutoTokenizer

model_name = "mistralai/Mistral-7B-v0.1"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 解决 pad_token 问题
tokenizer.pad_token = tokenizer.eos_token  

# Tokenize 影评数据
def preprocess_function(examples):
    encoding = tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=256
    )
    # 添加 labels，使其等于 input_ids（语言模型训练时通常这样做）
    encoding["labels"] = encoding["input_ids"].copy()
    return encoding

tokenized_datasets = dataset.map(preprocess_function, batched=True)

print(tokenized_datasets[0])

```

```python
import os
import torch
import pandas as pd
import numpy as np
from transformers import (
    AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer, EvalPrediction
)
from peft import LoraConfig, get_peft_model
from datasets import Dataset

model_name = "deepseek-ai/deepseek-coder-6.7b-instruct"

bnb_config = {
    "load_in_4bit": True,
    "bnb_4bit_compute_dtype": torch.float16,
    "bnb_4bit_use_double_quant": True,
    "bnb_4bit_quant_type": "nf4",
    "llm_int8_enable_fp32_cpu_offload": True
}

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    **bnb_config
)

tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token 

lora_config = LoraConfig(
    r=8,
    lora_alpha=32,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"]
)

model = get_peft_model(model, lora_config)

file_path = r"C:\Users\45713\Desktop\pythonneed\movie_review.txt"
with open(file_path, "r", encoding="utf-8") as f:
    raw_text = f.read()

reviews = [review.strip() for review in raw_text.split("=====") if review.strip()]
df = pd.DataFrame({"text": reviews})
dataset = Dataset.from_pandas(df)

def tokenize_function(examples):
    tokenized = tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=512
    )
    tokenized["labels"] = tokenized["input_ids"].copy()  
    return tokenized

tokenized_datasets = dataset.map(tokenize_function, batched=True)

class CustomTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        """
        兼容 `Trainer` 可能传入的额外参数（如 `num_items_in_batch`）。
        """
        labels = inputs.pop("labels")
        inputs["labels"] = labels  
        outputs = model(**inputs)
        logits = outputs.logits

        loss_fct = torch.nn.CrossEntropyLoss()
        shift_logits = logits[..., :-1, :].contiguous()
        shift_labels = labels[..., 1:].contiguous()
        loss = loss_fct(shift_logits.view(-1, shift_logits.size(-1)), shift_labels.view(-1))

        return (loss, outputs) if return_outputs else loss

def compute_metrics(eval_pred: EvalPrediction):
    logits, labels = eval_pred.predictions, eval_pred.label_ids
    logits = torch.tensor(logits)
    labels = torch.tensor(labels)

    loss_fct = torch.nn.CrossEntropyLoss()
    shift_logits = logits[..., :-1, :].contiguous()
    shift_labels = labels[..., 1:].contiguous()
    loss = loss_fct(shift_logits.view(-1, shift_logits.size(-1)), shift_labels.view(-1))

    print(f"📊 计算 eval_loss: {loss.item()}") 

    return {"eval_loss": loss.item()}  


output_dir = r"C:\Users\45713\Desktop\pythonneed\deepseek_lora_finetune"
training_args = TrainingArguments(
    output_dir=output_dir,
    evaluation_strategy="steps",  
    save_strategy="steps",
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=8,
    save_total_limit=2,
    learning_rate=2e-4,
    num_train_epochs=2,
    fp16=True,
    optim="adamw_bnb_8bit",
    report_to="none",
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss", 
    greater_is_better=False,  
    do_eval=True,  
    eval_steps=500,  
)   

trainer = CustomTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets,
    eval_dataset=tokenized_datasets,
    compute_metrics=compute_metrics
)

trainer.train()

trainer.save_model(output_dir)
print("🎉 训练完成，模型已保存至:", output_dir)

if os.path.exists(output_dir):
    print("📂 目录内容:", os.listdir(output_dir))
else:
    print("❌ 训练后模型未正确保存，请检查！")
```

```python
import torch
import os
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel, PeftConfig

# ✅ 设定路径
base_model_name = "deepseek-ai/deepseek-coder-6.7b-instruct"  # 原始模型
lora_model_path = r"C:\Users\45713\Desktop\pythonneed\deepseek_lora_finetune"  # LoRA 适配器路径

# ✅ 设定 `offload_dir`，确保目录存在
offload_dir = r"C:\Users\45713\Desktop\pythonneed\offload"
os.makedirs(offload_dir, exist_ok=True)

# ✅ 1. 加载 **原始 DeepSeek LLM 模型**
model = AutoModelForCausalLM.from_pretrained(
    base_model_name,
    torch_dtype=torch.float16,
    device_map="auto",  # ✅ 自动匹配 GPU / CPU
    offload_folder=offload_dir  # ✅ 设定 `offload_dir`
)

# ✅ 2. **加载 LoRA 配置**
lora_config = PeftConfig.from_pretrained(lora_model_path)

# ✅ 3. **使用 `PeftModel.from_pretrained()` 正确适配 LoRA**
try:
    model = PeftModel.from_pretrained(
        model,
        lora_model_path,
        is_trainable=False,  # ✅ 只进行推理
    )
except KeyError as e:
    print(f"❌ 加载 LoRA 适配器时出错: {e}")
    print("⚠️ 请检查 LoRA 适配器是否与 base_model 匹配！")

# ✅ 4. **合并 LoRA 权重**
model = model.merge_and_unload()  # ✅ 将 LoRA 权重合并到 `base_model`

# ✅ 让模型进入评估模式
model.eval()

# ✅ 重新加载 Tokenizer
tokenizer = AutoTokenizer.from_pretrained(base_model_name)

print("✅ LoRA 适配器加载完成！")

# ✅ 进行推理测试
prompt = "影评：这部电影的故事情节如何？"
inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

# 🚀 生成文本
with torch.no_grad():
    output = model.generate(**inputs, max_length=100)

result = tokenizer.decode(output[0], skip_special_tokens=True)
print("🎬 生成的影评分析：", result)
```
